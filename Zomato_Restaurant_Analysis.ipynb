# Train-test split

```python
#importing libraries
from sklearn.model_selection import train_test_split
```

```python
#Training and Testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

# Scale data if required

```python
#import libraries
from sklearn.preprocessing import StandardScaler
```

```python
# Redefine X and y using the encoded df
X = df[['Aggregate rating',
        'Votes',
        'Average Cost for two',
        'Has Online delivery',
        'Has Table booking']]
y = df['Price range']
```

```python
 #Re-perform train-test split to ensure X_train and X_test are updated
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)
```

```python
scaler = StandardScaler()

# Scale training data
X_train_scaled = scaler.fit_transform(X_train)

# Scale testing data
X_test_scaled = scaler.transform(X_test)
```

# Train classification models

```python
#import libraries
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
# Logistic Regression

# 2. Create Logistic Regression model
logistic_model = LogisticRegression(max_iter=1000)

# 3. Train the model
logistic_model.fit(X_train_scaled, y_train)
```

```python
#Decision Tree
decision_tree = DecisionTreeClassifier(random_state=42)
decision_tree.fit(X_train, y_train)
```

```python
# Random Forest
random_forest = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
random_forest.fit(X_train, y_train)
```

# Evaluate & compare models

```python
#import libraries
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
# Predictions
y_pred_logistic = logistic_model.predict(X_test_scaled)
y_pred_tree = decision_tree.predict(X_test)
y_pred_forest = random_forest.predict(X_test)
```

```python
# Function to evaluate model
def evaluate_model(name, y_test, y_pred):
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted', zero_division=0)
    recall = recall_score(y_test, y_pred, average='weighted', zero_division=0)
    f1 = f1_score(y_test, y_pred, average='weighted', zero_division=0)
    "\n", name
    "Accuracy :", round(accuracy, 4)
    "Precision:", round(precision, 4)
    "Recall   :", round(recall, 4)
    "F1 Score :", round(f1, 4)
    return accuracy
```

```python
# Evaluate all models
acc_logistic = evaluate_model(
    "Logistic Regression",
    y_test,
    y_pred_logistic
)
acc_tree = evaluate_model(
    "Decision Tree",
    y_test,
    y_pred_tree
)
acc_forest = evaluate_model(
    "Random Forest",
    y_test,
    y_pred_forest
)
```

```python
# Compare Accuracy
results = {
    "Logistic Regression": acc_logistic,
    "Decision Tree": acc_tree,
    "Random Forest": acc_forest
}
print("\nModel Comparison:")
for model, accuracy in results.items():
    print(model, ":", round(accuracy * 100, 2), "%")
```

```python
# Find best model
best_model = max(results, key=results.get)

print("\nBest Model:", best_model)
print("Best Accuracy:", round(results[best_model] * 100, 2), "%")
```
